```python
import numpy as np

def main():
    A = np.array([[0, 1], [1,1], [2,1], [3,1]])
    B = np.array([[-1], [0.2], [0.9], [2.1]])

    x = np.linalg.solve(A.T @ A, A.T @ B)
    print(x)

if __name__ == "__main__":
    main()
```

```python
from turtle import left

import matplotlib.pyplot as plt

def graph(x,y,ttl, position):
    plt.subplot(2,2,position)
    plt.xlabel('x Axis')
    plt.ylabel('y Axis')
    plt.title(ttl)
    left=0.0
    bottom=0.0
    right=1.4
    top=1.4
    wspace=0.4
    hspace=0.4
    plt.subplots_adjust(left=left, bottom=bottom, right=right, top=top, wspace=wspace, hspace=hspace)
    plt.plot(x,y)

def Linear_Equation():
    x = np.arange(-10, 10, 0.1)
    y = x
    graph(x,y,'y = x', 1)
    
def Quadratic_Equation():
    x = np.arange(-10, 10, 0.1)
    y = x**2
    graph(x,y,'y = x^2', 2)

def Trigonometry_Sine():
    x = np.arange(0, 2*np.pi, 0.1)
    y = np.sin(x)
    graph(x,y,'y = sin(x)', 3)
    plt.xticks(np.arange(0, 2*np.pi+0.1, np.pi/2), ['0', 'π/2', 'π', '3π/2', '2π'])


def Trigonometry_Cosine():
    x = np.arange(0, 2*np.pi, 0.1)
    y = np.cos(x)
    graph(x,y,'y = cos(x)', 4)
    plt.xticks(np.arange(0, 2*np.pi+0.1, np.pi/2), ['0', 'π/2', 'π', '3π/2', '2π'])

def execute():
    Linear_Equation()
    Quadratic_Equation()
    Trigonometry_Sine()
    Trigonometry_Cosine()


if __name__ == "__main__":
    execute()
```
