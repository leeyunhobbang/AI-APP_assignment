````python
import torch
import numpy as np

print(torch.__version__)
````

````python
def print_tensor_info(tensor, name):
    print(f"Tensor {name} :", tensor)
    print(f"Shape of {name} :", tensor.shape)
    print(f"Dimension of {name} :", tensor.dim())

def tensor_A():
    A = torch.FloatTensor([1, 2, 3, 4, 5])
    print_tensor_info(A, "A")
    print(f"Data type of A :", A.dtype, "\n")

def tensor_B():
    B = torch.FloatTensor([[1, 2, 3], [4, 5, 6]])
    print_tensor_info(B, "B")
    print(f"Number of elements in B :", B.numel(), "\n")

def tensor_C():
    C = torch.arange(1, 13, 1).view(2, 2, 3)
    print_tensor_info(C, "C")
    print(f"Number of elements in C :", C.numel(), "\n")


if __name__ == "__main__":
    tensor_A()
    tensor_B()
    tensor_C()
````

````python
def Tensor():
    A = torch.tensor([[1,2],[3,4]])
    B = torch.tensor([[5,6],[7,8]])
    return A,B

def Concat(tensor_1, tensor_2):
    C = torch.cat([tensor_1, tensor_2], dim=0)
    print(C, C.shape)
    
    D = torch.cat([tensor_1, tensor_2], dim=1)
    print(D, D.shape)

if __name__ == "__main__":
    A,B = Tensor()
    Concat(A,B)
````

# Q3. 
---
* 2-D tensor가 결합되는 방향의 차이로, dim = 0 은 row, dim = 1은 column방향으로 결합된다.
* 따라서 shape 또한 dim = 0 은 4 * 2로, dim = 1 은 2 * 4인 shape을 가진다.

````python
def stack(tensor_1, tensor_2):
    C = torch.stack([tensor_1, tensor_2], dim = 0)
    D = torch.stack([tensor_1, tensor_2], dim = 1)

    print(C, C.shape)
    print()
    print(D, D.shape)

if __name__ == "__main__":
    A ,B = Tensor()
    stack(A,B)
````

# Q8.
---
* tensor의 형태( _e.j._ 차원, rank, 축의 개수)는 바뀌지 않고 element-wise한 연산으로 진행됨
---

````python
def Multiplication():
    A = torch.tensor([
    [1., 2.],
    [3., 4.]
    ])

    B = torch.tensor([
        [5., 6.],
        [7., 8.]
    ])
    return A,B

def ElementWiseMultiplication(tensor_A, tensor_B):
    C = tensor_A * tensor_B
    print("Element-wise Multiplication")
    print("-" * 20)
    print(C)
    print("-" * 20)

def MatrixMultiplication_1(tensor_A, tensor_B):
    D = tensor_A @ tensor_B
    print("\n Matrix Multiplication")
    print("-" * 20)
    print(D)
    print("-" * 20)

def MatrixMultiplication_2(tensor_A, tensor_B):
    D = torch.matmul(tensor_A, tensor_B)
    print("\n Matrix Multiplication")
    print("-" * 20)
    print(D)
    print("-" * 20)


if __name__ == "__main__":
    A, B = Multiplication()
    ElementWiseMultiplication(A,B)
    MatrixMultiplication_1(A,B)
    MatrixMultiplication_2(A,B)

````

# Q9.

* *가 수행하는 연산은 element-wise한 곱셈(Hadamard Product)이므로, 같은 위치에 있는 원소끼리만 곱함
* @가 수행하는 연산은 행렬곱(Matrix Product)이므로, A•B는  A(i by j)의 row, B(j by k)의 column을 곱해 새로운 행렬은 만드는 연산임.

````python
def Transpose():
    A = torch.tensor([
        [1., 2., 3.],
        [4., 5., 6.]
    ])

    B = A.T

    print(A)
    print(A.shape)
    print()
    print(B)
    print(B.shape)

if __name__ == "__main__":
    Transpose()
````

````python
def StatisticalOperation():
    A = torch.tensor([
        [10., 20., 30.],
        [40., 50., 60.],
        [70., 80., 90.]
    ])
    return A

def PrintOperation(tensor_A):
    print(torch.sum(tensor_A))
    print(torch.mean(tensor_A))
    print(torch.max(tensor_A))
    print(torch.min(tensor_A))

if __name__ == "__main__":
    A = StatisticalOperation()
    PrintOperation(A)
````

````python
def DimensionOperation():
    A = torch.tensor([
        [10., 20., 30.],
        [40., 50., 60.],
        [70., 80., 90.]
    ])

    print(torch.sum(A, dim=0))
    print(torch.mean(A, dim=0))
    print()
    print(torch.sum(A, dim=1))
    print(torch.mean(A, dim=1))
    print("\n", "-" * 20, "\n")
    print(torch.sum(A, dim=0).shape)
    print(torch.mean(A, dim=0).shape)
    print()
    print(torch.sum(A, dim=1).shape)
    print(torch.mean(A, dim=1).shape)


if __name__ == "__main__":
    DimensionOperation()
````

# Q11.
---
* dim = 0 : 각 column 내부의 원소들끼리 연산됨
* dim = 1 : 각 row 내부의 원소들끼리 연산됨
* 합산 혹은 평균값을 도축하여 차원이 축소되므로 shape은 서로 동일함
---

````python
def LinkUpPractice():
    # 1. 크기가 (3, 4)인 Tensor A 생성
    A = torch.arange(1.,13.,1).view(3,4) 
    # 자료형 미지정시, A.type은 integer이기에 평균 계산 시 오류 발생하므로 float로 생성

    print(f"Tensor A:\n{A}\n")
    print(f"A shape: {A.shape}\n")


    # 2. Tensor A를 전치하여 Tensor B 생성
    B = A.T

    print(f"Tensor B (A의 전치):\n{B}\n")
    print(f"B shape: {B.shape}\n")


    # 3. Tensor A와 B를 행렬 곱셈하여 Tensor C 생성
    C = A @ B

    print(f"Tensor C (A @ B):\n{C}\n")
    print(f"C shape: {C.shape}\n")


    # 4. Tensor C의 각 행에 대해 평균을 계산하여 Tensor D에 저장
    D = torch.mean(C, dim = 1)

    print(f"Tensor D (C의 행별 평균):\n{D}\n")
    print(f"D shape: {D.shape}\n")

    # D = torch.mean(C)
    # print(f"Tensor D (C의 평균):\n{D}\n")
    # print(f"D shape: {D.shape}\n")


    # 5. Tensor D의 모든 원소의 합을 계산
    final_sum = torch.sum(D)

    print(f"최종 결과값: {final_sum}\n")

if __name__ == "__main__":
    LinkUpPractice()
````
